# Checking for Missing Values

Missing values can reduce model performance and may require imputation.

This section calculates the number and percentage of missing values in each feature.

```python
missing = df.isnull().sum()

missing_percentage = (missing / len(df)) * 100

missing_table = pd.DataFrame({
    "Missing Values": missing,
    "Percentage": missing_percentage
})

missing_table
```

# Checking for Duplicate Records

Duplicate observations may bias machine learning models.

This section identifies duplicate rows within the dataset.

```python
duplicates = df.duplicated().sum()

print("Duplicate Rows:", duplicates)
```

```python
# Remove duplicates if they exist

df = df.drop_duplicates()

print("New Shape:", df.shape)
```

# Column Names

Reviewing the feature names helps us understand the information available for prediction.

These variables will later be categorized as:

- Numerical Features
- Categorical Features
- Target Variables

```python
df.columns
```

# Understanding Categorical Variables

Machine learning algorithms require numerical input.

Before encoding categorical variables, it is important to inspect their unique values and identify inconsistencies such as:

- Different spellings
- Capitalization differences
- Unexpected categories

```python
categorical_columns = [
    "Industry",
    "Country",
    "Ransomware_Group",
    "Attack_Vector",
    "Payment_Status",
    "Data_Exfiltration"
]

for column in categorical_columns:
    print("=" * 50)
    print(column)
    print(df[column].unique())
```

# Frequency Distribution of Categories

The frequency of each category provides insight into the distribution of ransomware attacks across industries, countries, and attack vectors.

This information will later be visualized during exploratory data analysis.

```python
for column in categorical_columns:
    print("\n")
    print("=" * 50)
    print(column)
    print(df[column].value_counts())
```

# Converting Date Features

The `Attack_Date` column is converted into the datetime format to enable time-based analysis.

This will allow future exploration of:

- Monthly attack trends
- Annual attack trends
- Seasonal patterns

```python
df["Attack_Date"] = pd.to_datetime(df["Attack_Date"])

df.info()
```

# Data Types After Conversion

The dataset is inspected once more to verify that the date conversion was successful.

```python
df.dtypes
```

# Numerical and Categorical Features

Separating feature types simplifies preprocessing during model development.

```python
numerical_columns = df.select_dtypes(include=['int64','float64']).columns

categorical_columns = df.select_dtypes(include=['object']).columns

print("Numerical Features")
print(numerical_columns)

print("\nCategorical Features")
print(categorical_columns)
```

# Saving the Clean Dataset

After completing the initial inspection, the cleaned dataset is saved for use in subsequent notebooks.
